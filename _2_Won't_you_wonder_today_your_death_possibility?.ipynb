```python
import numpy as np

class program :

  def __init__ (self):
    self.human_name = input("input your name ~ ^^ ♪ ")
    self.possibility = []

  def death_possiblity(self):
    self.a, self.b, self.c, self.d, self.e = np.random.rand(5)
    self.possibility.extend([self.a, self.b, self.c, self.d, self.e])
    print("🆚", self.possibility)

    self.k = np.argmax(self.possibility)
    self.f = self.possibility[self.k]
    self.your_destiny = self.f / sum(self.possibility)
    return self.your_destiny

  def destiny(self):

    rand = np.random.rand()

    if self.your_destiny > rand:
      print("😒 you die")
      print("1. possibility", self.your_destiny)
      print("2. rand", rand)

    else :
      print("😉 you live")
      print("1. possibility", self.your_destiny)
      print("2. rand", rand)


program = program()
result = []

for i in range(10):
  program.death_possiblity()
  program.destiny()
  result.append(program.your_destiny)
```
