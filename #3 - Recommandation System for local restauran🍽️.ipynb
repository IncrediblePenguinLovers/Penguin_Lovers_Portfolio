```python
""" → 각 메뉴 종류

1. 떡볶이
2. 참치마요
3. 계란라면
4. 꼬마김밥 (5줄)
5. 오뎅탕
6. 닭볶음밥
7. 우동

→ 1인 세트 메뉴

1. 떡볶이 + 꼬마김밥 (3줄)
2. 라면 + 꼬마김밥 (3줄)
3. 우동 + 꼬마김밥 (3줄)

"""

"""
♪ 메뉴를 추천해주는 프로그램
"""

import numpy as np
import random

class Recommandation :

  def __init__ (self) :
    self.For_1 = {1 : '떡볶이', 2 : '참치마요', 3 : '계란라면', 4 : '꼬마김밥', 5 : '오뎅탕', 6 : '닭볶음밥', 7 : '우동'}
    self.For_2 = {1 : '떡볶이 + 꼬마김밥', 2 : '라면 + 꼬마김밥', 3 : '우동 + 꼬마김밥'}

  def purchase (self):
    # 1인분을 원하면 각 메뉴 종류 중 한 가지를 선택
    # 1인 세트를 원하면 1인 세트 메뉴 종류 중 한 가지를 선택
    Want = int(input(" 당신은 메뉴 하나를 원하시나요? 당신은 1인 세트를 원하시나요? → 메뉴 하나 : 0, 1인분 : 1"))
    if Want == 0 :
      a = int(np.random.choice(np.arange(1,8), 1))
      print(a)
      print(self.For_1[a])

    elif Want == 1 :
      b = int(np.random.choice(np.arange(1,4), 1))
      print(b)
      print(self.For_2[b])

human = Recommandation()
human.purchase()
```
